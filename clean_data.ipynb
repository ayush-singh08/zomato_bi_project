### Removing irrelevant and high-noise features like URLs, text-heavy columns (reviews, menu items) and redundant columns to focus on structured attributes that impact restaurant performance.

```python
columns_to_drop = ['url','phone','address','reviews_list', 'menu_item', 'location','rest_type']
df.drop(columns=columns_to_drop, inplace=True)
```

### Renaming columns for better understandings

```python
df.rename(columns={
    'listed_in(type)': 'restaurant_type',
    'approx_cost(for two people)' : 'cost_for_two',
    'rate' : 'rating',
    'listed_in(city)' : 'location'
}, inplace=True)
```

### Changing data types 
#### 1. 'rating' column 
it is in format (X/5) string type we will make it float type with X only

```python
#dropping columns with NaN or some other value

df = df[df['rating'].notnull()]              # remove NaN
df = df[~df['rating'].isin(['NEW', '-'])]   # remove invalid strings
```

```python
#Split the string by '/', select the first part and then convert it to float type 

df['rating'] = df['rating'].str.split('/').str[0].astype(float)
```

```python
df.head(3)
```

```python
df['cost_for_two'].unique()
```

### Cleaning the cost_for_two column 
1. removing null valued rows from dataframe
2. removing ',' from the values
3. changing string to float

```python
df = df[df['cost_for_two'].notnull()]
```

```python
df['cost_for_two'] = df['cost_for_two'].str.replace(',', '')
```

```python
df['cost_for_two'] = df['cost_for_two'].astype(float)
```

```python
df['cost_for_two'].unique()
```

```python
# feature engineering 

# 1. 
df['cost_category'] = pd.cut(
    df['cost_for_two'],
    bins=[0, 500, 1500, 10000],
    labels=['Budget', 'Mid-Range', 'Premium']
)
```

```python
# 2.

df['rating_category'] = pd.cut(
    df['rating'],
    bins=[0, 3, 4, 5],
    labels=['Poor', 'Good', 'Excellent']
)
```

```python
#Standardize Yes/No Columns

df['online_order'] = df['online_order'].str.strip()
df['book_table'] = df['book_table'].str.strip()
```

```python
#Fill Missing Values

df['dish_liked'] = df['dish_liked'].fillna('Not Specified')
df['cuisines'] = df['cuisines'].fillna('Unknown')
```
